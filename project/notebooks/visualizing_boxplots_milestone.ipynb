## Why Boxplots Matter

Boxplots are essential in exploratory data analysis because they:
- Provide a compact summary of data distribution
- Make outliers immediately visible
- Allow easy comparison across multiple variables
- Complement histograms by highlighting quartiles clearly
- Help identify spread and central tendency at a glance

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

print("✓ Libraries imported successfully")
```

## 1. Understanding Boxplot Components

A boxplot consists of several key components:

### The Box (Interquartile Range - IQR)
- **Bottom edge**: Q1 (25th percentile)
- **Top edge**: Q3 (75th percentile)  
- **Box height**: IQR = Q3 - Q1 (contains middle 50% of data)

### The Median Line
- Line inside the box showing the middle value (50th percentile)
- Different from the mean (average)

### The Whiskers
- Extend from box to 1.5 × IQR
- **Lower whisker**: Q1 - 1.5 × IQR (but not below minimum data point)
- **Upper whisker**: Q3 + 1.5 × IQR (but not above maximum data point)

### Outliers
- Individual points beyond the whiskers
- Potential anomalies or extreme values
- **Important**: Not always errors - need context!

## 2. Create Sample Data

Let's create a sample dataset with multiple numeric columns and intentional outliers for demonstration.

```python
# Set random seed for reproducibility
np.random.seed(42)

# Create sample dataset with different distributions
data = {
    'Age': np.concatenate([
        np.random.normal(35, 10, 95),  # Normal distribution
        [75, 80, 15]  # Some outliers
    ]),
    'Income': np.concatenate([
        np.random.normal(50000, 15000, 90),
        [150000, 180000, 200000, 220000, 10000]  # High earners and one low outlier
    ]),
    'Experience_Years': np.concatenate([
        np.random.normal(8, 4, 92),
        [25, 28, 1]  # Long experience and novice outliers
    ]),
    'Satisfaction_Score': np.concatenate([
        np.random.normal(7.5, 1.2, 93),
        [3.0, 3.5, 10.0, 2.0]  # Low and high satisfaction outliers
    ]),
    'Hours_Worked_Weekly': np.concatenate([
        np.random.normal(40, 5, 94),
        [70, 75, 15, 18]  # Overworkers and part-timers
    ])
}

df = pd.DataFrame(data)

print(f"Dataset created with {len(df)} rows and {len(df.columns)} columns")
print(f"\nColumns: {list(df.columns)}")
print("\nFirst few rows:")
df.head()
```

```python
# Display basic statistics
print("Descriptive Statistics:")
df.describe()
```

## 3. Creating a Single-Column Boxplot

Let's start by creating a boxplot for one column to understand the distribution.

```python
# Select a column to analyze
column = 'Age'

# Calculate key statistics
q1 = df[column].quantile(0.25)
median = df[column].median()
q3 = df[column].quantile(0.75)
iqr = q3 - q1
lower_whisker = q1 - 1.5 * iqr
upper_whisker = q3 + 1.5 * iqr

print(f"Statistics for {column}:")
print(f"  Median (Q2):        {median:.2f}")
print(f"  Q1 (25th %ile):     {q1:.2f}")
print(f"  Q3 (75th %ile):     {q3:.2f}")
print(f"  IQR (Q3 - Q1):      {iqr:.2f}")
print(f"  Lower Whisker:      {lower_whisker:.2f}")
print(f"  Upper Whisker:      {upper_whisker:.2f}")

# Identify outliers
outliers = df[(df[column] < lower_whisker) | (df[column] > upper_whisker)][column]
print(f"\n  Number of outliers: {len(outliers)}")
if len(outliers) > 0:
    print(f"  Outlier values:     {sorted(outliers.values)}")
```

```python
# Create the boxplot
plt.figure(figsize=(8, 6))

plt.boxplot(df[column].dropna(), 
            vert=True, 
            patch_artist=True,
            boxprops=dict(facecolor='lightblue', edgecolor='black'),
            medianprops=dict(color='red', linewidth=2),
            whiskerprops=dict(color='black', linewidth=1.5),
            capprops=dict(color='black', linewidth=1.5),
            flierprops=dict(marker='o', markerfacecolor='red', markersize=8, alpha=0.5))

plt.ylabel(column, fontsize=12)
plt.title(f'Boxplot of {column}', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)

# Add median annotation
plt.text(1.15, median, f'Median: {median:.2f}', 
         verticalalignment='center', fontsize=10, color='red')

plt.tight_layout()
plt.show()

print("\n✓ Boxplot created successfully")
```

### Interpretation

From the boxplot above:
1. The **red line** shows the median age
2. The **box** contains the middle 50% of ages
3. **Red dots** represent outliers (unusually high or low ages)
4. The **whiskers** extend to the extreme values within 1.5×IQR

## 4. Comparing Multiple Columns with Boxplots

One of the greatest strengths of boxplots is the ability to compare distributions side-by-side.

```python
# Select columns to compare
columns_to_compare = ['Satisfaction_Score', 'Experience_Years', 'Hours_Worked_Weekly']

# Display comparative statistics
print("Comparative Statistics:")
print("=" * 80)
comparison_stats = pd.DataFrame({
    'Median': df[columns_to_compare].median(),
    'Q1': df[columns_to_compare].quantile(0.25),
    'Q3': df[columns_to_compare].quantile(0.75),
    'IQR': df[columns_to_compare].quantile(0.75) - df[columns_to_compare].quantile(0.25),
    'Range': df[columns_to_compare].max() - df[columns_to_compare].min()
})
print(comparison_stats)
```

```python
# Create side-by-side boxplots
plt.figure(figsize=(14, 6))

positions = range(1, len(columns_to_compare) + 1)
bp_data = [df[col].dropna() for col in columns_to_compare]

bp = plt.boxplot(bp_data, 
                 positions=positions, 
                 patch_artist=True,
                 labels=columns_to_compare,
                 boxprops=dict(facecolor='lightgreen', edgecolor='black'),
                 medianprops=dict(color='darkred', linewidth=2),
                 whiskerprops=dict(color='black', linewidth=1.5),
                 capprops=dict(color='black', linewidth=1.5),
                 flierprops=dict(marker='D', markerfacecolor='red', 
                               markersize=6, alpha=0.6))

plt.xlabel('Variables', fontsize=12, fontweight='bold')
plt.ylabel('Values', fontsize=12, fontweight='bold')
plt.title('Comparing Distributions Across Multiple Columns', 
          fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ Comparison boxplot created")
```

```python
# Analyze key insights from comparison
print("\n" + "=" * 80)
print("KEY INSIGHTS FROM COMPARISON:")
print("=" * 80)

# Most variable column
largest_iqr_col = comparison_stats['IQR'].idxmax()
print(f"  • Most variable column: {largest_iqr_col}")
print(f"    (IQR: {comparison_stats.loc[largest_iqr_col, 'IQR']:.2f})")

# Least variable column
smallest_iqr_col = comparison_stats['IQR'].idxmin()
print(f"\n  • Least variable column: {smallest_iqr_col}")
print(f"    (IQR: {comparison_stats.loc[smallest_iqr_col, 'IQR']:.2f})")

# Count outliers for each column
print("\n  Outlier counts:")
for col in columns_to_compare:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outlier_count = len(df[(df[col] < lower) | (df[col] > upper)])
    print(f"    - {col}: {outlier_count} outliers")
```

## 5. Standardized Boxplots for Different Scales

When comparing columns with vastly different scales (e.g., Age vs Income), standardization helps.

```python
# Select columns with different scales
columns_different_scales = ['Age', 'Income']

# Standardize data (z-score normalization)
df_standardized = df[columns_different_scales].apply(lambda x: (x - x.mean()) / x.std())

print("Standardization (Z-Score):")
print("Z-score = (value - mean) / standard_deviation")
print("\nThis transforms data to have mean=0 and std=1")
print("\nStandardized statistics:")
print(df_standardized.describe())
```

```python
# Create standardized boxplots
plt.figure(figsize=(14, 6))

positions = range(1, len(columns_different_scales) + 1)
bp_data = [df_standardized[col].dropna() for col in columns_different_scales]

bp = plt.boxplot(bp_data, 
                 positions=positions, 
                 patch_artist=True,
                 labels=columns_different_scales,
                 boxprops=dict(facecolor='lightcoral', edgecolor='black'),
                 medianprops=dict(color='darkblue', linewidth=2),
                 whiskerprops=dict(color='black', linewidth=1.5),
                 capprops=dict(color='black', linewidth=1.5),
                 flierprops=dict(marker='s', markerfacecolor='blue', 
                               markersize=6, alpha=0.6))

plt.axhline(y=0, color='gray', linestyle='--', alpha=0.7, linewidth=1)
plt.xlabel('Variables', fontsize=12, fontweight='bold')
plt.ylabel('Z-Score (Standard Deviations from Mean)', fontsize=12, fontweight='bold')
plt.title('Standardized Boxplots for Fair Comparison', 
          fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ Standardized boxplot allows fair comparison despite different scales")
```

## 6. Outlier Detection and Interpretation

**CRITICAL**: Outliers are not always errors!

```python
# Detailed outlier analysis for Income
column = 'Income'

q1 = df[column].quantile(0.25)
q3 = df[column].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# Identify outliers
outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

print("=" * 80)
print(f"OUTLIER DETECTION: {column}")
print("=" * 80)
print(f"\nTotal data points:     {len(df)}")
print(f"Number of outliers:    {len(outliers)}")
print(f"Percentage outliers:   {(len(outliers) / len(df) * 100):.2f}%")
print(f"\nLower bound:           ${lower_bound:,.2f}")
print(f"Upper bound:           ${upper_bound:,.2f}")

if len(outliers) > 0:
    print(f"\nOutlier values:")
    for idx, value in enumerate(sorted(outliers[column].values), 1):
        print(f"  {idx}. ${value:,.2f}")
```

### How to Interpret Outliers

Outliers can represent:

#### 1. Data Entry Errors
- Typos (e.g., age = 150 instead of 15)
- Wrong units (e.g., meters vs. centimeters)
- **Action**: Verify and correct if confirmed as errors

#### 2. Genuine Extreme Values
- High earners in income data
- Athletes in physical performance data
- **Action**: Keep them! They represent real phenomena

#### 3. Different Populations
- Part-time vs. full-time workers
- Different age groups mixed together
- **Action**: Consider separate analysis or grouping

#### 4. Measurement Errors
- Instrument malfunction
- Recording mistakes
- **Action**: Investigate root cause

### Before Removing Outliers:
- ✓ Understand the domain context
- ✓ Investigate the cause
- ✓ Document your decision
- ✓ Consider separate analysis with/without outliers

## 7. Boxplot vs Histogram

Boxplots and histograms provide complementary insights.

```python
column = 'Hours_Worked_Weekly'

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot
axes[0].boxplot(df[column].dropna(), 
                vert=True, 
                patch_artist=True,
                boxprops=dict(facecolor='lightblue', edgecolor='black'),
                medianprops=dict(color='red', linewidth=2),
                flierprops=dict(marker='o', markerfacecolor='red', markersize=8, alpha=0.5))
axes[0].set_ylabel(column, fontsize=11)
axes[0].set_title('Boxplot', fontsize=12, fontweight='bold')
axes[0].grid(axis='y', alpha=0.3)

# Histogram
axes[1].hist(df[column].dropna(), bins=20, color='lightgreen', 
             edgecolor='black', alpha=0.7)
axes[1].axvline(df[column].median(), color='red', linestyle='--', 
                linewidth=2, label='Median')
axes[1].axvline(df[column].mean(), color='blue', linestyle='--', 
                linewidth=2, label='Mean')
axes[1].set_xlabel(column, fontsize=11)
axes[1].set_ylabel('Frequency', fontsize=11)
axes[1].set_title('Histogram', fontsize=12, fontweight='bold')
axes[1].legend()
axes[1].grid(axis='y', alpha=0.3)

plt.suptitle(f'Boxplot vs Histogram: {column}', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("\n" + "=" * 80)
print("BOXPLOT STRENGTHS:")
print("  • Quick outlier identification")
print("  • Easy comparison across groups")
print("  • Clear quartile visualization")
print("  • Compact representation")
print("\nHISTOGRAM STRENGTHS:")
print("  • Shows distribution shape (normal, skewed, bimodal)")
print("  • Frequency information")
print("  • Detailed view of data density")
print("  • Reveals modality (number of peaks)")
print("=" * 80)
```

## 8. Using Seaborn for Enhanced Boxplots

Seaborn provides additional boxplot features and better styling.

```python
# Create boxplots for all numeric columns using seaborn
plt.figure(figsize=(14, 6))

# Melt the dataframe for seaborn
df_melted = df.melt(var_name='Variable', value_name='Value')

sns.boxplot(data=df_melted, x='Variable', y='Value', palette='Set2')
plt.xticks(rotation=45, ha='right')
plt.xlabel('Variables', fontsize=12, fontweight='bold')
plt.ylabel('Values', fontsize=12, fontweight='bold')
plt.title('Seaborn Boxplot: All Numeric Columns', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ Seaborn provides cleaner styling and easier syntax")
```
