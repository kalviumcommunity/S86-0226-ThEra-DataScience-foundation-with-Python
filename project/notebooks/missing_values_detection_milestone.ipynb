## Section 1: Understanding Missing Values

### What Are Missing Values?

Missing values represent the **ABSENCE** of data:
- **NaN** (Not a Number) - Pandas/NumPy standard
- **None** - Python's null object (converted to NaN in Pandas)
- Empty strings `""` - sometimes used incorrectly
- Placeholder values - like -999, "N/A" (domain-specific)

Pandas standardizes all missing values as **NaN** for consistency.

### Creating Sample Data with Missing Values

````python
# Create a simple DataFrame with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, np.nan, 30, 28, np.nan],
    'Salary': [50000, 55000, np.nan, 60000, 52000],
    'Department': ['Sales', 'IT', None, 'Sales', 'IT']
}

df_example = pd.DataFrame(data)
print("Sample DataFrame with missing values:")
df_example
````

**Observations:**
- NaN appears in Age (rows 1, 4) and Salary (row 2)
- None appears in Department (row 2) - displayed as NaN
- Name column has no missing values

## Section 2: Detecting Missing Values with isnull()

### What is isnull()?

`isnull()` returns a **boolean DataFrame**:
- **True** where values are missing (NaN)
- **False** where values exist

**Syntax:** `df.isnull()` or `df.isna()`  
**Note:** `isnull()` and `isna()` are identical

````python
# Detect missing values - returns boolean DataFrame
print("Boolean mask showing where values are missing:")
print("(True = Missing, False = Exists)")
print()
df_example.isnull()
````

### Interpretation:
- **True** = Missing value detected at that location
- **False** = Value exists at that location
- Each cell shows if that specific position has missing data

````python
# Check specific column for missing values
print("Missing values in Age column:")
print(df_example['Age'].isnull())
````

### Using notnull() - Opposite of isnull()

`notnull()` returns:
- **True** where values exist
- **False** where values are missing

````python
# Detect existing values
print("Boolean mask showing where values exist:")
print("(True = Exists, False = Missing)")
print()
df_example.notnull()
````

---

## Section 3: Counting Missing Values

### Count Missing Values Per Column

Combining `isnull()` with `sum()` counts missing values:
- `isnull()` creates True/False mask
- `sum()` treats True as 1, False as 0
- Result: count of missing values per column

````python
# Count missing values per column
missing_counts = df_example.isnull().sum()
print("Missing value counts per column:")
print(missing_counts)
````

````python
# Total missing values in entire DataFrame
total_missing = df_example.isnull().sum().sum()
print(f"Total missing values in entire DataFrame: {total_missing}")
````

````python
# Count non-missing (existing) values per column
print("Non-missing value counts:")
print(df_example.count())
print()
print("Alternative using notnull().sum():")
print(df_example.notnull().sum())
````

---

## Section 4: Identifying Columns with Missing Data

### Find Columns with ANY Missing Values

Use `any()` to check if column has **at least one** missing value

````python
# Check which columns have any missing values
columns_with_missing = df_example.isnull().any()
print("Columns with at least one missing value:")
print(columns_with_missing)
print()

# Get list of column names with missing data
missing_column_names = columns_with_missing[columns_with_missing].index.tolist()
print(f"Column names with missing data: {missing_column_names}")
````

### Find Columns with ALL Missing Values

Use `all()` to check if column is **completely empty**

````python
# Check for completely empty columns
print("Columns with all values missing:")
print(df_example.isnull().all())
print()

# Create example with completely empty column
df_with_empty = df_example.copy()
df_with_empty['EmptyColumn'] = np.nan

print("After adding completely empty column:")
print(df_with_empty.isnull().all())
````

---

## Section 5: Inspecting Rows with Missing Data

### Find Rows with ANY Missing Values

Use `axis=1` to check **across columns** (row-wise)

````python
# Check which rows have any missing values
rows_with_missing = df_example.isnull().any(axis=1)
print("Rows with at least one missing value:")
print(rows_with_missing)
print()
print(f"Total rows with missing data: {rows_with_missing.sum()} out of {len(df_example)}")
````

### View Only Rows with Missing Data

Filter DataFrame to show only affected rows

````python
# Filter to show only rows containing missing values
df_missing_rows = df_example[df_example.isnull().any(axis=1)]
print("Rows containing missing values:")
df_missing_rows
````

### View Only Complete Rows (No Missing Values)

````python
# Filter to show only complete rows
df_complete_rows = df_example[df_example.notnull().all(axis=1)]
print("Complete rows (no missing values):")
df_complete_rows
````

## Section 6: Creating Comprehensive Missing Values Summary

### Build a Complete Summary Table

A comprehensive summary should include:
- Column name
- Data type
- Total values
- Missing values count
- Missing percentage

````python
def missing_values_summary(df):
    """
    Generate comprehensive missing value report.
    
    Parameters:
    -----------
    df : pandas.DataFrame
        The DataFrame to analyze
    
    Returns:
    --------
    pandas.DataFrame
        Summary with missing value statistics
    """
    total = df.shape[0]
    missing_count = df.isnull().sum()
    missing_percent = (missing_count / total) * 100
    data_types = df.dtypes
    
    summary = pd.DataFrame({
        'Column': df.columns,
        'Data_Type': data_types.values,
        'Total_Values': total,
        'Missing_Count': missing_count.values,
        'Missing_Percent': missing_percent.values
    })
    
    # Sort by missing count descending
    summary = summary.sort_values('Missing_Count', ascending=False)
    summary = summary.reset_index(drop=True)
    
    return summary

print("✓ Function defined: missing_values_summary()")
````

````python
# Create comprehensive summary
summary = missing_values_summary(df_example)
print("Comprehensive Missing Value Summary:")
summary
````

````python
# Filter to show only columns with missing values
problematic_columns = summary[summary['Missing_Count'] > 0]
print("Columns with missing data (problematic):")
problematic_columns
````

## Section 7: Real-World Example - Student Records Dataset

Let's apply what we've learned to a more realistic dataset

````python
# Create larger, more realistic student dataset
np.random.seed(42)
n_students = 50

student_data = {
    'StudentID': range(1001, 1001 + n_students),
    'Name': [f'Student_{i}' for i in range(1, n_students + 1)],
    'Age': np.random.randint(18, 25, n_students),
    'Gender': np.random.choice(['M', 'F', 'Other', np.nan], n_students, p=[0.45, 0.45, 0.05, 0.05]),
    'Major': np.random.choice(['CS', 'Math', 'Physics', 'Biology', np.nan], n_students, p=[0.3, 0.25, 0.2, 0.2, 0.05]),
    'GPA': np.round(np.random.uniform(2.0, 4.0, n_students), 2),
    'Midterm_Score': np.random.randint(60, 100, n_students),
    'Final_Score': np.random.choice([*np.random.randint(65, 100, 45), *[np.nan]*5]),
    'Attendance_Percent': np.round(np.random.uniform(70, 100, n_students), 1),
    'Email': [f'student{i}@university.edu' if i % 7 != 0 else np.nan for i in range(1, n_students + 1)],
    'Phone': [f'555-{1000+i:04d}' if i % 5 != 0 else np.nan for i in range(1, n_students + 1)]
}

df_students = pd.DataFrame(student_data)
print(f"✓ Created student dataset with {len(df_students)} students and {len(df_students.columns)} columns")
print()
print("Preview:")
df_students.head(10)
````

````python
# Quick missing values check
print("Quick Missing Values Check:")
print("=" * 50)
print(f"Total missing values: {df_students.isnull().sum().sum()}")
print()
print("Missing values per column:")
print(df_students.isnull().sum())
````

````python
# Detailed analysis
summary_students = missing_values_summary(df_students)
print("Detailed Missing Value Summary:")
summary_students
````

````python
# Identify high-priority issues
high_missing = summary_students[summary_students['Missing_Percent'] > 5]
print("⚠️  Columns with > 5% missing data:")
if not high_missing.empty:
    print(high_missing[['Column', 'Missing_Count', 'Missing_Percent']])
    print()
    print("These columns need attention before analysis!")
else:
    print("✓ No columns have excessive missing data (>5%)")
````

````python
# Sample rows with missing data
rows_with_missing = df_students[df_students.isnull().any(axis=1)]
print(f"Found {len(rows_with_missing)} rows with missing values out of {len(df_students)} total")
print(f"Percentage: {len(rows_with_missing)/len(df_students)*100:.1f}%")
print()
print("First 5 rows with missing data:")
rows_with_missing.head()
````

## Section 8: Best Practices

### ✓ DO

1. **Always check for missing values immediately after loading data**
   ```python
   df = pd.read_csv('data.csv')
   print(df.isnull().sum())  # First thing!
   ```

2. **Create comprehensive summaries before analysis**
   ```python
   summary = missing_values_summary(df)
   print(summary)
   ```

3. **Inspect rows with missing data**
   ```python
   df_missing = df[df.isnull().any(axis=1)]
   print(df_missing.head(20))
   ```

4. **Document your findings**
   - Record which columns have missing data
   - Note percentages and patterns
   - Keep in data quality log

5. **Detection before action**
   - Never fill or drop without detecting first
   - Understand the problem before solving

### ✗ AVOID

1. **Ignoring missing values**
   - Leads to incorrect statistics
   - Causes silent errors

2. **Assuming data is complete**
   - Real-world data always has gaps
   - Check even "clean" datasets

3. **Confusing missing with zero**
   - NaN ≠ 0
   - Missing means "unknown"

4. **Skipping row inspection**
   - Patterns reveal collection issues
   - Context helps decision-making

5. **Acting without analysis**
   - Don't immediately drop or fill
   - Understand severity first

## Section 9: Recommended Workflow

Follow this workflow for any new dataset:

````python
# Example workflow with our student dataset
print("MISSING VALUE DETECTION WORKFLOW")
print("=" * 60)

# Step 1: Load data (already done above)
print("\n1. Data loaded ✓")

# Step 2: Quick overview
print("\n2. Quick overview:")
print(f"   Shape: {df_students.shape}")
print(f"   Columns: {list(df_students.columns)}")

# Step 3: Check for missing values
print("\n3. Missing values per column:")
print(df_students.isnull().sum())

# Step 4: Create detailed summary
print("\n4. Detailed summary:")
summary = missing_values_summary(df_students)
print(summary.to_string())

# Step 5: Identify problematic columns
print("\n5. Problematic columns (>5% missing):")
problematic = summary[summary['Missing_Percent'] > 5]
if not problematic.empty:
    print(problematic[['Column', 'Missing_Count', 'Missing_Percent']].to_string())
else:
    print("   None - all columns <5% missing ✓")

# Step 6: Check affected rows
print("\n6. Rows affected:")
rows_affected = df_students.isnull().any(axis=1).sum()
print(f"   {rows_affected}/{len(df_students)} rows have missing data ({rows_affected/len(df_students)*100:.1f}%)")

# Step 7: Document and decide
print("\n7. Next steps:")
print("   - Document findings")
print("   - Determine handling strategy")
print("   - Proceed with appropriate cleaning")

print("\n" + "=" * 60)
print("✓ WORKFLOW COMPLETE")
````
